### Cleaning of Dataset

```python
import pandas as pd
import numpy as np
from configs.system import project_dir
```

```python
p_dir = project_dir()
data_dir = p_dir.RAW_DIR
clean_dir = p_dir.PROCESSED_DIR
dataset_path = data_dir/'ChargingRecords.csv'
```

```python
## loading the dataset
df = pd.read_csv(dataset_path)
df.sample(4)
```

```python
df.info()
```

```python
### fixing the date time columns
df['StartDay'] = pd.to_datetime(df['StartDay']).dt.date
df['EndDay'] = pd.to_datetime(df['EndDay']).dt.date
df['StartDatetime'] = pd.to_datetime(df['StartDatetime'])
df['EndDatetime'] = pd.to_datetime(df['EndDatetime'])
```

```python
df['StartTime'] = pd.to_datetime(df['StartTime']).dt.time
df['EndTime'] = pd.to_datetime(df['EndTime']).dt.time
```

```python
df.sample(4)
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```
