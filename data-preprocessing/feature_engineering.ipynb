### Feature Engineering

```python
import pandas as pd
import numpy as np
from configs.system import project_dir
```

```python
p_dir = project_dir()
clean_dir = p_dir.PROCESSED_DIR
dataset_path = clean_dir/'cleaned_dataset.csv'
```

```python
df = pd.read_csv(dataset_path,index_col=0)
df.sample(4)
```

```python
df['StartMonth'] = pd.to_datetime(df['StartDay']).dt.month
df['StartWeek'] = pd.to_datetime(df['StartDay']).dt.dayofweek

df['EndMonth'] = pd.to_datetime(df['EndDay']).dt.month
df['EndWeek'] = pd.to_datetime(df['EndDay']).dt.dayofweek
```

```python
df.sample(3)
```

```python
df['DemandPerMonth'] = df.groupby('StartMonth')['Demand'].transform('sum')
```

```python
df.sample(2)
```

```python
df.groupby('StartMonth')['DemandPerMonth'].count()
```

```python
df.groupby('StartMonth')['DemandPerMonth'].sum().sort_values(ascending=False)
```

```python
df.groupby('StartMonth')['Demand'].sum().sort_values(ascending=False)
```

```python
df.columns
```

```python
df['StartDayOfWeek'] = pd.to_datetime(df['StartDay']).dt.day_name()
```

```python
df.sample(3)
```

```python
df['DemandPerDay'] = df.groupby('StartDayOfWeek')['Demand'].transform('mean').round(2)
```

```python
df.sample(3)
```

```python
df.groupby('StartDayOfWeek')['DemandPerDay'].mean()
```

```python
df.groupby('StartDayOfWeek')['DemandPerDay'].sum().round(2).sort_values(ascending=False)
```

```python
feature_dir = p_dir.FEATURED_DIR
```

```python
from datetime import time

df['StartTime'] = pd.to_datetime(df['StartTime'],format='%H:%M:%S').dt.time

df['Shift'] = df['StartTime'].apply(
    lambda x: 'Day' if x < time(14,0) else 'Night'
)
```

```python
df.groupby('Shift')['Demand'].mean().round(2)
```

```python
df.groupby('Shift')['Demand'].count()
```

```python
df.to_csv(feature_dir/'featured_dataset.csv')
```
